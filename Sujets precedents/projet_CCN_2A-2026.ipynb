# Fonctionnement

<img src="https://github.com/PlantecJY/python-mic/raw/master/rdn.png" alt="Drawing" style="width: 200px;"/>

La figure ci-dessus schématise un exemple très simple de réseau de neurones artificiels. Le réseau est constitué de neurones reliés entre eux et figurés par des cercles de couleur.

Un neurone de base (cercle bleu) possède plusieurs entrées par lesquelles il reçoit un signal issu d'autres neurones et plusieurs sorties par lesquelles il peut sous certaines conditions émettre un signal vers d'autres neurones. 

Certains neurones – les neurones d'entrée (cercles verts) – sont activés en recevant une information directement depuis l'extérieur. Les neurones de sortie (ici, un seul cercle jaune), quant à eux, ne sont pas nécessairement du même nombre que les neurones d'entrée ; on s'intéresse seulement à leur état, car il ne retransmettent pas d'information.

Le neurone de base (cercle bleu) est donc conçu comme un automate doté d'une fonction de transfert qui transforme selon des règles précises ses entrées notées i en une sortie notée j. Par exemple, le premier neurone bleu de la colonne de gauche transforme deux entrées en un signal de sortie distribué vers les 5 neurones de la colonne de droite. 

Dans le modèle élémentaire que nous allons mettre en œuvre, l'information qui parcourt une entrée vaut 0 (pas de signal) ou 1 (un signal). Le neurone effectue une somme pondérée (une combinaison linéaire en quelque sorte) des entrées, chaque entrée étant affectée d'un poids w (on parle de « poids synaptique »). Chaque neurone se comporte donc différemment des autres ; l'efficacité de la transmission des signaux d'un neurone à l'autre peut varier. La somme pondérée étant effectuée, chaque neurone compare cette somme à une valeur seuil : si cette somme est supérieure ou égale à ce seuil, le neurone s'active et émet un signal vers les autres neurones suivants ; dans le cas contraire, il reste inactif. Ceci est bien entendu un modèle ultra-simplifié du fonctionnement d'un neurone biologique.

Le seuil de chaque neurone ainsi que la valeur des poids synaptiques wij sont donc des paramètres importants du réseau : selon les informations d'entrée (dans ce TP, ce seront des ensembles de 0 et de 1 - cf ci-dessous), les neurones de sortie seront activés ou pas. Le réseau est donc un outil permettant d'opérer une certaine transformation (on parlera de classification) des patterns (un ensemble d'entrées) qui lui sont soumis en un pattern de sortie (dans l'exemple ci-dessus, une seule valeur de sortie) ; notons que plusieurs patterns d'entrée différents peuvent conduire au même pattern de sortie.

L'idée du réseau de neurones est de choisir ces poids et ce seuil de telle façon que la transformation opérée par le réseau conduise à un pattern de de sortie aussi proche que possible de celle voulue par le concepteur. C’est ce qu'on nomme la phase d’apprentissage du réseau. Pour un réseau de neurones formels, apprendre revient donc à déterminer le jeu de coefficients synaptiques permettant de classifier les données présentées ; lorsque ce jeu est déterminé, il est attendu du réseau qu'il classifie correctement une entrée nouvelle (ne faisant pas partie des données qui ont servi à l'apprentissage).

    Dans un modèle plus riche, le neurone fonctionne avec des nombres réels (souvent compris dans l’intervalle [0,1] ou [-1,1]).

## Création du réseau de neurones

Le réseau de neurones que l'on va utiliser est un réseau extrêmement simple : il possède 30 entrées et 10 sorties. Chacune des 30 entrées est reliée aux 10 sorties. Le réseau contient donc 300 neurones et est donc complètement représenté par la matrice 30x10 des poids de chaque neurone.

Dans cette partie, on crée une fonction nommée *init_poids* qui va créer cette matrice contenant les poids du réseau de neurones. Elle accepte deux paramètres qui sont :

La fonction *init_poids* réalise les étapes suivantes :
- créer une matrice poids de dimension *dimEntree* x *dimSortie* ;

```python
from numpy import random


def init_poids(dimEntree, dimSortie):
    random.seed(2) # NE PAS MODIFIER CETTE LIGNE QUI REND DETERMINISTE LE RESULTAT
    
    poids = []
    for i in range(dimEntree):
        col = []
        for j in range(dimSortie):
            col.append(random.random()*100-50)
        poids.append(col)
    return poids

### COMMENCEZ VOTRE CODE ICI ###

    
```

## Apprentissage
L'apprentissage consiste à ajuster les poids de manière à ce que les sorties du réseau de neurones (calculées par *calcul_reseau*) correspondent bien aux chiffres fournis en entrée du réseau. Cet apprentissage est réalisé itérativement. A chaque itération on va faire progresser les poids, en confrontant les représentations des chiffres du dictionnaire *d* aux sorties calculées. Le nombre d'itérations est ici fixé.

Ici, cette apprentissage s'appuie sur trois fonctions, correspondant aux boucles suivantes :

une boucle sur les itérations : fonction (*apprendre*)
-> à chaque itération, une boucle sur toutes les entrées du dictionnaire : fonction (*apprendre_reseau*)
-> à chaque entrée, une boucle sur tous les neurones : fonction (*apprendre_neurone*)

Voici d'abord, la fonction nommée *apprendre_neurone* qui permet à partir d'une entrée *e* représentant un chiffre d'ajuster la matrice *poids* de telle manière que la sortie calculée se rapproche de *1* si le chiffre *e* représente *j* ou de *0* sinon.

Cet apprentissage se base sur la formule suivante que nous vous fournissons, qui ajuste le poids du neurone situé entre la i-ième entrée et la j-ième sortie :

    poids[i][j] = poids[i][j] + (valeur_desiree - valeur_calculee) * e[i] * h

La variable *h* est un paramètre d'apprentissage, positif, que nous vous proposons de fixer à *0.05* et que vous pouvez faire varier.

Optimiser la matrice *poids* de telle façon que les sorties calculées la fonction *calcul_reseau* correspondent aux chiffres du dictionnaire.

Exercice : initialiser une matrice *poids* de dimension 30x10 ; afficher (à nouveau) les poids des neurones connectés à la sortie j=3 ; lancer la fonction *apprendre_neurone* avec en entrée la première valeur du chiffre 9 dans le dictionnaire d (9 sera la sortie attendue) et pour le neurone j=3 ; réafficher les poids des neurones connectés à la sortie j=3 après actualisation par la fonction *apprendre_neurone*.

```python
def apprendre_neurone(e, poids, j, sortie_attendue):
    # pas : vitesse d'apprentissage du perceptron, caractérise la convergence ou non de valeur_calculee vers valeur_desiree
    # En effet, une vitesse trop élevée (h = 5) amplifiera l'erreur
    h = 0.05
    # Calcul de la sortie du neurone j, avec l'entree e et la matrice de poids
    valeur_calculee = calcul_neurone(j,e,poids)
    # Boucle sur les composantes de e et ajustement des poids correspondant avec la formule donnée
    for i in range(len(e)):
        # Calcul de la valeur desirée en fonction de la sortie attendue
        valeur_desiree = 0
        if sortie_attendue == j:
            valeur_desiree = 1
        # Actualisation du poids[i][j] (cf. formule précédente)
        # L'erreur est représentée par: valeur_desiree - valeur_calculee
        poids[i][j] = poids[i][j] + (valeur_desiree - valeur_calculee)*e[i]*h 
    return poids    

### COMMENCEZ VOTRE CODE ICI ###
```
